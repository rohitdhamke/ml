```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
data = pd.read_csv('/workspaces/ml/KNN/House Price Prediction Dataset (1).csv')
data
```

```python
data = data.drop(['Id'], axis=1)
data
```

```python
x =  data.drop(["Price", "Location", "Condition", "Garage"], axis=1)
```

```python
y = data["Price"]
```

```python
columns = data.columns
for col in columns:
    print(data[col].value_counts())
    print("\n")
```

```python
numerical  = ['Area','Bedrooms','Bathrooms','Floors', 'YearBuilt']
```

```python
ct = ColumnTransformer(
    transformers = [("num", StandardScaler(), numerical)]
)
```

```python
x_scaled = ct.fit_transform(x)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size = 0.2, random_state = 32)
```

```python
knnr = KNeighborsRegressor(n_neighbors = 550, metric = "euclidean")
```

```python
knnr = KNeighborsRegressor(n_neighbors = 650, metric = "euclidean")
knnr.fit(X_train, Y_train)
pred = knnr.predict(X_test)
score = r2_score(Y_test, pred)
print(score)
```

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ====================== advantages of seaborn =====================

# better color combination
# plot direct from the dataframe
# advanced plot(heatmap, kdeplor(key dencity), violinplot, boxplot)
# it has some sample datasets to practice
```

```python
penguin_data = sns.load_dataset("penguins")
penguin_data
```

```python
# scatterplot , stripplot , swarmplot 

# color palette = colorblind , dark , deep , bright
```

```python
# sns.scatterplot(data=penguin_data, x="body_mass_g", y="flipper_length_mm", palette="colorblind", hue="sex")

# sns.scatterplot(data=penguin_data, x="body_mass_g", y="flipper_length_mm", palette="deep", hue="sex")
# sns.scatterplot(data=penguin_data, x="body_mass_g", y="flipper_length_mm", palette="dark", hue="sex")
sns.scatterplot(data=penguin_data, x="bill_length_mm", y="bill_depth_mm", palette="bright", hue="species")
```

```python
sns.stripplot(data=penguin_data, x="island", y="body_mass_g", palette="colorblind", hue="species")
```

```python
sns.stripplot(data=penguin_data, x="sex", y="body_mass_g", palette="colorblind", hue="species")
```

```python
plt.figure(figsize=(12,13))
sns.swarmplot(data=penguin_data, x="sex", y="body_mass_g", palette="colorblind", hue="species")
```

```python
plt.figure(figsize=(7,8))
sns.swarmplot(data=penguin_data, x="bill_length_mm", y="body_mass_g", palette="colorblind", hue="species")
```

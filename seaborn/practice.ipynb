```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
# 1 Create a scatter plot showing the relationship between:

# X-axis → Age
# Y-axis → Fare

# Requirements:

# hue = Sex
# palette = "deep"
# Use sns.set_style("whitegrid")

sns.set_style("whitegrid")
sns.scatterplot(data=titanic, x="age", y="fare", hue="sex", palette="deep")
```

```python
# 2 Create a scatter plot.

# X → Age
# Y → Fare

# Requirements:

# hue = Survived
# palette = "bright"

# Observe whether passengers paying higher fares had better survival rates.

sns.scatterplot(data=titanic, x="age", y="fare", hue="survived", palette="bright")
```

```python
# 3 Create a strip plot.

# X → Pclass
# Y → Fare

# Requirements:

# hue = Sex
# palette = "colorblind"

sns.stripplot(data=titanic, x="pclass", y="fare", hue="sex", palette="colorblind")
```

```python
# 4 Create a strip plot.

# X → Survived
# Y → Age

# Requirements:

# hue = Sex
# palette = "Set2"

sns.stripplot(data=titanic, x="survived", y="age", palette="Set2", hue="sex")
```

```python
# 5 Create a swarm plot.

# X → Sex
# Y → Age

# Requirements:

# hue = Survived
# palette = "dark"

sns.swarmplot(data=titanic, x="sex", y="age", hue="survived", palette="dark")
```

```python
# 6 Create a swarm plot.

# X → Pclass
# Y → Fare

# Requirements:

# hue = Survived
# palette = "pastel"

sns.swarmplot(data=titanic, x="pclass", y="fare", hue="survived", palette="pastel")
```

```python
# 7 Create a box plot.

# Y → Age

# Requirements:

# hue = Survived
# palette = "coolwarm"

# Find:

# Median age
# Outliers
# Age spread

sns.boxplot(data=titanic, y="age", hue="survived", palette="coolwarm")
```

```python
# 8 Create a box plot.

# Y → Fare

# Requirements:

# hue = Pclass
# palette = "viridis"

sns.boxplot(data=titanic, y="fare", hue="pclass", palette="viridis")
```

```python
# 9 Create a KDE plot.

# X → Age

# Requirements:

# hue = Sex
# Fill the density
# palette = "Set1"

sns.kdeplot(data=titanic, x="age", hue="sex", palette="Set1")
```

```python
# 10 Create a KDE plot.

# X → Fare

# Requirements:

# hue = Survived
# Fill the density
# palette = "tab10"

sns.kdeplot(data=titanic, y="fare", hue="survived", palette="tab10")
```

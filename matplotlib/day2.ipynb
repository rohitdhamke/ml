```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(32)
array1 = np.random.normal(455, 3.4,100) #median, std, number of values(size)
array2 = np.random.normal(455,6.7,100)
array3 = np.random.normal(467, 2.3, 100)
```

```python
 # Width = 8 inch
 # Height = 7 inch

plt.figure(figsize = (8,7))
plt.boxplot(
    [array1,array2,array3],
    showmeans=True,
    tick_labels=["Science","English","Maths"]
)
```

```python
labels = ["Science", "English", "Maths"]
marks = [45,67,78]
fig, axes = plt.subplots(nrows = 4, ncols = 1, figsize = (8,10))
axes[0].boxplot([array1, array2, array3], showmeans = True, tick_labels = ["Science", "English", "Maths"])
axes[1].plot(array1, array2, color = "green", marker = "d", mfc = "red", mec = "black")
axes[2].scatter(array1, array2, color = "yellow", marker = "h")
axes[3].barh(labels, marks, height =0.7, color = "green", edgecolor = "black")
plt.tight_layout()
```

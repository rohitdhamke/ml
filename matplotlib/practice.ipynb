```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Requirements:

# Blue line
# Dashed (--) linestyle
# Circle markers
# Marker size = 10
# Line width = 2
# Title = "Sales Growth"
# X label = "Months"
# Y label = "Sales"
```

```python
# 1

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]

plt.plot(x,y, color="blue", ls="dashed", lw="2", marker="o", ms='10' )
plt.title("Sales Growth")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# 2

# Requirements:

# Red line
# Diamond (d) marker
# Marker face color = yellow
# Marker edge color = black
# Dash-dot (-.) line
# Title = "Weekly Temperature"

days = [1,2,3,4,5,6,7]
temperature = [30,31,29,33,35,36,34]

plt.plot(days,temperature, "r-.", mfc="yellow", mec="black")
plt.title("Weekly Temperature")
plt.xlabel("days")
plt.ylabel("temperature")
plt.show()
```

```python
# 3

# Requirements:

# Green points
# Size = 120
# Black edge color
# Title = "Age vs Salary"

age = [18,20,22,24,26,28]
salary = [15000,18000,25000,28000,35000,40000]

plt.scatter(age,salary, color="green", s=120, edgecolor="black")
plt.title("age vs salary")
plt.xlabel("age")
plt.ylabel("salary")
plt.show()
```

```python
# 4

# Requirements:

# Sky blue bars
# Black border
# Width = 0.5
# Title = "Student Marks"

subjects = ["Math","Science","English","IT"]
marks = [78,84,72,91]

plt.bar(subjects,marks, color="skyblue", edgecolor="black", width=0.5)
plt.title("student marks")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.show()
```

```python
# 5

# Requirements:

# Orange bars
# Edge color = black
# Width = 0.7
# X-axis = City
# Y-axis = Population (Millions)

cities = ["Mumbai","Pune","Nagpur","Nashik"]
population = [20,7,3,2]

plt.bar(cities,population, color="orange", edgecolor="black", width=0.7)
plt.xlabel("city")
plt.ylabel("population (millions)")
plt.show()
```

```python
# # 6  

# Requirements:

# Different colors
# Shadow enabled
# Black border
# Explode Python slice only

languages = ["Python","Java","C++","JavaScript"]
students = [45,20,15,20]

explode = [0.3,0,0,0]
color=["skyblue", "orange", "green", "yellow"]

plt.pie(students, labels=languages, explode=explode, colors=color, shadow="True", wedgeprops={"edgecolor" : "black"} )
```

```python
#  7

# Requirements:

# Different colors
# Shadow = True
# Explode Rent slice
# Black border

expenses = ["Food","Rent","Travel","Shopping"]
money = [12000,18000,5000,7000]

explode=[0,0.4,0,0]
color=["red", "green", "blue", "yellow"]

plt.pie(money, labels=expenses, shadow="True", explode=explode, colors=color, wedgeprops={"edgecolor" : "black"})
```

```python
#  8

# Requirements:

# Green line
# Square (s) marker
# Marker size = 8
# Marker face color = yellow
# Marker edge color = black
# Dashed line
# Line width = 3
# Title = "Square Numbers"

x = np.arange(1,11)
y = x**2

plt.plot(x,y, color="green", marker="s", ms=8, mfc="yellow", mec="black", ls="dashed", lw=3)
plt.title("square numbers")
plt.show()
```

```python
#  9

# Create all 4 graphs in one program

# Make

# Line Plot
# Scatter Plot
# Bar Chart
# Pie Chart

# Each graph should have:

# Title
# Appropriate colors
# Labels (where applicable)
# Black edges wherever possible



months = ["Jan","Feb","Mar","Apr","May"]
sales = [25,40,35,50,60]


# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [25, 40, 35, 50, 60]

# line plot
plt.plot(months,sales, color="red")
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.show()

# scatter plot 
plt.scatter(months,sales, color="green", marker="d", s=120, edgecolors="black")
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.show()

# bar 
plt.bar(months,sales, color="skyblue", width=0.7, edgecolor="black")
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.show()

# pie chart
color=["red","green","orange","cyan","yellow"]
plt.pie(sales,labels=months, colors=color, shadow="True", wedgeprops={"edgecolor" : "black"})
plt.title("monthly sales")
plt.show()
```

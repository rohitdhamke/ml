```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# matplotlib is a plotting tool that is used to understand the pattern in the dataset visually.

# ls=["-.", "dotted", "dashed" , "solid"]
# marker = ["o", "h", "d", "p", "s"]
# mfc = marker face color
# ms = marker size 
# 
```

```python
x = np.array([3,4,6,7])
y = np.array([1,2,3,4])
plt.plot(x,y, ls="dotted", color="red", lw="2.5", marker="o", ms=13 , mfc="black", mec="#077A9C") 
plt.title("My first line plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# scatter shows only marker

plt.scatter(x,y, color="#808000", s=40, edgecolors="black")
```

```python
sectors = ["Agriculture", "Industry", "Medical"]
employement = [40,30,30]
color= ["green" , "gray", "red"]
explode= [0.4,0,0]

plt.pie(employement, labels=sectors, colors=color, shadow="True", explode=explode, wedgeprops = {"edgecolor" : "black"})
```

```python
plt.bar(sectors,employement , color=["orange","white","green"],edgecolor="black", width=0.6)
```

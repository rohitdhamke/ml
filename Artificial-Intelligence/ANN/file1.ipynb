```python
# ANN = Artificial Neural Network
#================================

# What is ANN?
#=============
# ANN stands for Artificial Neural Network.
# It is a Deep Learning model inspired by the structure of biological neurons in the human brain.
# ANN learns patterns from data by adjusting weights and biases during training.
```

```python
# Artificial Neuron:
# An artificial neuron takes inputs, multiplies them by weights, adds bias, and applies an activation function.

# ann has three types of layes:
# 1. input layer (Accepts input features)
# 2. hidden layer (Learns patterns from input)
# 3. output layer (Produces final prediction)
```

```python
# example: 

# Inputs:
# - Hours studied
# - Attendance
# - Previous marks

# Output:
# - Pass = 1
# - Fail = 0
```

```python
# ===============  Activation Function  ================
#=======================================================

# activation function weighted sum ko transform krta hai.
# Without activation functions, multiple neural network layers 
# with only linear operations would still behave like a single linear transformation.

#======  formula  =================

#1. ReLU (Rectified Linear Unit):
# formula:
#         f(z) = max(0,z)

# rules :
#       if  z > 0 , output is z
#       if  z < 0 , output is 0 
```

```python
# ========== Forward Propagation ==================
#===========================================================

# Forward Propagation means input se output tak information pass karna.

# process : 1.input ==> 2.weighted sum ==> 3.ReLU ==> 4.Neuron output

# This entire process is called Forward Propagation.
```

```python
#===========  implementation  =================
#==============================================


#   z  is weighted sum .



# inputs
x1 = 2
x2 = 3

# weights
w1 = 0.5
w2 = 0.8

# bias:
b = 0.2

# weighted sum (z):
# formula for weighted sum: """ z=x1w1 + x2w2 + b """

# z = 2*0.5 + 3*0.8 + 0.2
# print("weighted sum is : ",z)
# Weighted Sum
z = x1*w1 + x2*w2 + b

# ReLU Activation
output = max(0, z)

print("Weighted Sum:", z)
print("Neuron Output( Activation function ):", output)
```

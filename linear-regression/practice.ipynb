```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5])
y = np.array([4,5,6,8,10])

mean_x = x.mean()
meab_y = y.mean()
print(mean_x)
print(meab_y)
```

```python
numer = 0
denom = 0

for i in range(len(x)):
    numer += (x[i] - mean_x)*(y[i] - meab_y)
    denom += (x[i] - mean_x) ** 2
print(numer)
print(denom)
```

```python
m = numer / denom
print("slope is :", m)
b = meab_y - m * mean_x
print("intersept is :", b)
```

```python
# linear regression prediction

y_prediction = 1.5 * x + 2.1
y_prediction
```

```python
plt.scatter(x,y, color="green")
plt.plot(x,y_prediction, color="red")
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
x_2d = x.reshape(len(x),1)
lr.fit(x_2d,y)
y_pred = lr.predict(x_2d)
# print(y_pred)
# print(y)
plt.scatter(x,y, color="green")
plt.plot(x,y_pred, color="red")
```

```python
#  Question 2 ------>

bikeage = np.array([1,2,3,4,5,6,7,8])
milage = np.array([55,53,50,48,45,43,40,38])

print(bikeage)
print(milage)
```

```python
bikeage
milage

bikeage_mean = bikeage.mean()
milage_mean = milage.mean()
print(bikeage_mean)
print(milage_mean)
```

```python
numeretor = 0 
denometor = 0

for i in range(len(bikeage)):
    numeretor += (bikeage[i] - bikeage_mean) * (milage[i] - milage_mean)
    denometor += (bikeage[i] - bikeage_mean) ** 2
print(numeretor)
print(denometor)
```

```python
m = numeretor / denometor
print("slope :",m)
```

```python
b = milage_mean - m * bikeage_mean
print(b)
```

```python
y_milage_pred = m * bikeage + b
print(y_milage_pred)

plt.scatter(bikeage,milage, color="green")
plt.plot(bikeage,y_milage_pred, color="red")
```

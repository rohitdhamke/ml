```python
mean_x = x.mean()
mean_y = y.mean()
len(y)
```

```python
number = 0 # counter
denom = 0
for i in range(len(x)):
    number += (x[i] - mean_x) * (y[i] - mean_y)               #(5 - 3) * (5 - 4) = 2
    denom += (x[i] - mean_x) ** 2                       #(5 - 3)^2 = 4
m = number/denom                      #0.6
b = mean_y  - (m * mean_x)            #2.2
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
hours_studies = np.array([1,2,3,4,5])
marks = np.array([2,4,5,4,5])
x = hours_studies.reshape(len(x),1)
x.ndim
y = marks.copy()
```

```python
lr.fit(x,y)
```

```python
m = lr.coef_
```

```python
b = lr.intercept_
```

```python
y_pred = m * x + b
# print(f"According to model : {y_pred}")
# print(y)
```

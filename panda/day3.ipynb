```python
import re
# re = regular expression / regex
# regular expresssion is a library that used to clean text data.
#  It is used to find patterns in text data and replace them with something else.
```

```python
# Ranges
# [A-Z] : Capital A to Z
# [a-z]: Small a to z
# [0-9]: Digits from 0 to 9
# Special ranges
# \w : Digits alphabets(Capital asweel as small)
# \d: Digits
# \s: Space
# \S: Non-space characters
# Quantifiers:
# {3,4}:Range quantifiers
# ?: It detects till last and second last character of a word given before it
# + : One or more than one times
# * : Zero or more than zero times
# Logical:
# |: or operator
# ^ : not operator
# Delimeter
# \
```

```python
# ===================     text cleaning     ======================
```

```python
def clean_text(text):
    text = re.sub(r"http?:\/{2}\s+", "", text)  # remove http links
    text = re.sub(r"@\w+", "", text) # remove mentions
    text = re.sub(r"[^\w\s]", "", text) # remove punctuation 
    return text
data_second["Cleaned_text"] = data_second["message to examine"].apply(clean_text)

data_second
```

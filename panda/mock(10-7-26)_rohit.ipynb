```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv(r"/workspaces/ml/panda/sample_-_superstore (1).csv",encoding="latin-1")
data
```

```python
# 1
profit_data=data[data['Profit'] > 0]
profit_data
```

```python
#  2
customer_profit = profit_data.groupby('Customer Name')['Profit'].sum()
customer_profit

customer_orders = profit_data.groupby('Customer Name')['Order ID'].count()
customer_orders
```

```python
# 2

customer_profit = profit_data.groupby('Customer Name')['Profit'].sum()
customer_orders = profit_data.groupby('Customer Name')['Order ID'].count()
fil_customers = customer_profit[customer_orders > 5]
fil_customers.head(10)
```

```python
# 3
avg_dis = data.groupby('Category')['Discount'].mean()
fil_avg_dis = avg_dis[avg_dis > 0.15].sort_values(ascending=False)
fil_avg_dis
```

```python
# 4

region_quantity = data[data["Sales"] > 500].groupby('Region')["Quantity"].sum()
sorted_region_quan = region_quantity.sort_values(ascending=False)
sorted_region_quan
```

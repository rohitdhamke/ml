```python
data = pd.read_csv(r"/workspaces/ml/panda/sample_-_superstore (1).csv",encoding="latin-1")
data
```

```python
data.info()
```

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"] , format= "%d-%m-%Y") # if year has four digite then use "Y" if two digite then use "y"
data["Ship Date"] = pd.to_datetime(data["Ship Date"] , format= "%d-%m-%Y")
```

```python
data.info()
```

```python
data["No of Days"] = data["Ship Date"] - data["Order_Date"]
data
```

```python
data["No of Days"] = (data["Ship Date"] - data["Order_Date"]).dt.days
data
```

```python
data["State"]
```

```python
new_data = data.groupby("State").agg(Avg_days = ("No of Days" , "mean")).reset_index()
new_data
# avg_days = data.groupby("State")["No of Days"].mean().reset_index(name="avg_days")
# avg_days
```

```python
new_data["Avg_days"] = new_data["Avg_days"].astype(int)
new_data.sort_values("Avg_days", ascending=False)
```

```python
data["Year"] = data["Order_Date"].dt.year
data
```

```python
# yearly_sales = data.groupby("Year")["Sales"].sum().reset_index()
# yearly_sales.sort_values("Sales", ascending=False).reset_index()

yearly_sales = data.groupby("Year").agg(Total_Sales = ("Sales", "sum")).reset_index()
yearly_sales
yearly_profit = data.groupby("Year").agg(Total_profit = ("Profit", "sum")).reset_index()
yearly_profit
```

```python
yearly_sales.sort_values("Total_Sales", ascending= False)
yearly_profit.sort_values("Total_profit", ascending= False)
```

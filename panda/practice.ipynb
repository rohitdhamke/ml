```python
import pandas as pd
```

```python
data = pd.read_csv(r"/workspaces/ml/panda/sample_-_superstore (1).csv",encoding="latin-1")
data
```

```python
# 1

orders = data[data["Profit"] > 0]
orders
states = orders.groupby("State")["Sales"].sum().sort_values(ascending=False).head(5)
states
```

```python
data
```

```python
# 2

customer_profit = data.groupby("Customer Name")["Profit"].sum()
customer_profit
customer_orders = data.groupby("Customer Name")["Order ID"].count()
customer_orders

filtered_customer = customer_profit[customer_orders > 5]
filtered_customer.head(10)
```

```python
# 3 

mean_dis = data.groupby("Category")["Discount"].mean()
mean_dis

avg_dis = mean_dis[mean_dis > 0.15]
avg_dis
```

```python
# 4 

region = data[data["Sales"] > 500].groupby("Region")["Quantity"].sum()
region
region = region.sort_values(ascending=False)
region
```

```python
# 5

high_avg = data[data["Category"] == "Technology"]
high_avg = high_avg.groupby("Sub-Category")["Sales"].mean().sort_values(ascending=False).head(3)
high_avg
```

```python
data
```

```python
# 6

product_total = data.groupby("Product Name")["Profit"].sum()
product_total
product_quant = data.groupby("Product Name")["Quantity"].sum()
product_quant
filtered = product_total[product_quant > 10].sort_values(ascending=True)
filtered.head(5).reset_index()
```

```python
# 7

discount = data[data["Discount"] > 0.20]
discount
total = discount.groupby("Ship Mode")[["Sales", "Profit"]].sum()
total.sort_values(by="Profit", ascending=False).reset_index()
```

```python
# 8

customer = data.groupby("Customer Name")[["Sales", "Profit"]].sum()
customer[customer["Sales"] > 10000]
customer.sort_values(by="Sales", ascending=False).reset_index()
```

```python
# 9

top = data.groupby(["Region", "State"])["Sales"].sum()
top.sort_values(ascending=False)
```

```python
# 10 

cust = data.groupby(["Customer Name", "Category"])["Sales"].sum().reset_index()
cust
result = cust.sort_values(by=["Category", "Sales"], ascending=[True,False])
result.reset_index()
```

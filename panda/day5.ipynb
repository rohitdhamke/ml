```python
import pandas as pd
```

```python
data = pd.read_csv(r"/workspaces/ml/panda/sample_-_superstore (1).csv",encoding="latin-1")
data
```

```python
# calculate the avg_no_days it took a order to reach the shipment according to the shipmode.
```

```python
data.info()
```

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"] , format="%d-%m-%Y")
data["Ship Date"] = pd.to_datetime(data["Ship Date"] , format="%d-%m-%Y")
```

```python
data.info()
```

```python
data["no_of_days"] = (data["Ship Date"] - data["Order_Date"]).dt.days
data
```

```python
new_data = data.groupby("Ship Mode").agg(Avg_Days = ("no_of_days", "mean")).reset_index()
new_data
```

```python
new_data["Avg_Days"] = new_data["Avg_Days"].astype(int)
new_data
```

```python
data["Year"] = data["Order_Date"].dt.year
data
```

```python
# According to the year calculate how much total_profit the store got in each year
```

```python
profit = data.groupby("Year").agg(Total_pro = ("Profit", "sum")).reset_index()
profit
```
